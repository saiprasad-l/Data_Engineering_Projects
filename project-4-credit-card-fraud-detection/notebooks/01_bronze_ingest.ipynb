```python
# Bronze notebook: ingest raw CSV into Delta (paste into a Databricks Python notebook)
from pyspark.sql.functions import current_timestamp

# Update this path to the actual uploaded file path shown in Databricks UI
credit_csv = "dbfs:/Volumes/workspace/default/project_4/creditcard.csv"

# Read CSV with schema inference (small file ok)
raw = spark.read.option("header", True).option("inferSchema", True).csv(credit_csv)

# Add ingest timestamp
raw = raw.withColumn("ingest_ts", current_timestamp())

# Create database and delta path
spark.sql("CREATE DATABASE IF NOT EXISTS fraud_proj")
bronze_path = "dbfs:/Volumes/workspace/default/project_4/bronze/creditcard_raw"

raw.write.format("delta").mode("overwrite").save(bronze_path)

# Read back without registering a table
bronze_df = spark.read.format("delta").load(bronze_path)
display(bronze_df.limit(5))

# Save as Delta
# raw.write.format("delta").mode("overwrite").option("overwriteSchema","true").saveAsTable("fraud_proj.bronze_creditcard")
#spark.sql(f"CREATE TABLE IF NOT EXISTS fraud_proj.bronze_creditcard USING DELTA LOCATION '{bronze_path}'")



# display(spark.table("fraud_proj.bronze_creditcard").limit(5))
```
